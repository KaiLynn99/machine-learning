```python
# for vs 컴프리헨션{} 예제

sqare_dict = {}
for val in range(1, 11): # range(a, b) : a ~ (b-1)
  sqare_dict[val] = val**2 # val*val # key : 1 ~ 10, value : 1**2 ~ 10**2

print(sqare_dict)

sqare_dict = {val:val**2 for val in range(1, 11)}
print(sqare_dict)

# 컴프리헨션 {}셋
sqare_set = {val**2 for val in range(1, 11)}
print(sqare_set) # set은 순서가 없다

# 컴프리헨션 ()튜플
sqare_tuple = (val**2 for val in range(1, 11))
print(tuple(sqare_tuple))
```

```python
# 함수
# 최소/최대 정규화 함수(min max scaler) : (값 - 최소값) / (최대값-최소값) : 모든 연속된 숫자값이 0 ~ 1 스케일링

def min_max_scale(x, minnum, maxnum): # 값, 최소값, 최대값 순서로 입력
  return (x - minnum) / (maxnum - minnum) # 0~1
```

```python
my_score = 99
min = 50
max = 100

min_max_scale(my_score, min, max)
```

```python
other_score = 70

min_max_scale(other_score, min, max)
```

```python
# 단위가 달라도 최소/최대 정규화값 동일하다 -> 기계가 값은 값으로 가중치를 준다
large_my_score = 7000
large_min = 5000
large_max = 10000

min_max_scale(large_my_score, large_min, large_max)
```

```python
# 파이썬 람다문법
# 함수를 식으로 표현하는 방법. 일회성 함수(익명 함수) 생성할때 많이 사용하는 문법
# def 키워드를 사용해 함수를 선언할 필요가 없다.

# def min_max_scale(x, minnum, maxnum):
#   return (x - minnum) / (maxnum - minnum) # 0~1
# min_max_scale함수를 람다식으로 표현

min_max_scale2 = lambda x, minnum, maxnum : (x - minnum) / (maxnum - minnum)
```

```python
my_score = 99
min = 50
max = 100

min_max_scale2(other_score, min, max)
```

```python
# 고차함수(Higher-Order Funtion)
# 함수가 다른 함수의 인자로 입력되는 문법

# 1급함수(First-Class Funtion)
# 함수를 변수에 저장가능한 함수문법
kkk = min_max_scale # kkk변수안에 min_max_scaler 함수가저장 -> 1급함수
round(kkk(80, 60, 95), 3)
```

```python
# 콜백함수
# 함수가 다른 함수의 인자로 입력되어 고차함수 그 함수안에서 무조건 호출이 되는게 아니고
# 조건(이벤트)이 일치하면 호출하는 형태

def apply_fun(other_fun, value): # other_fun 함수변수(1급함수) -> 인자값(고차함수)
  result = other_fun(value) # 콜백함수
  result += value
  return result

def square(x):
  return x*x # x**2


apply_fun(square, 3) # 3**2 + 3
```

```python
# np.array([1, 10, 100]) > 10
# array([False, False,  True]) -> array(['거짓', '거짓',  '참'])

arr = np.array([1, 10, 100])

# 파이썬 반복문 사용
result = []
for value in arr:
  if value > 10:
    print(value)
    result.append('참') # result 배열이었다  result = np.append(result, '참')
  else:
    print(value)
    result.append('거짓')
print(result)
```

```python
# 파이썬 컴프리헨션 문법을 사용
result2 = ['참' if x>10 else '거짓' for x in arr]
result2
```

```python
# np.where(조건, 참일때, 거짓일때)
result3 = np.where(arr>10, '참', '거짓')
result3
```

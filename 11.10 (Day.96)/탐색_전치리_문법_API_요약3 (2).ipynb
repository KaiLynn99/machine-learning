```python
# 고차함수 : 함수의 인자로 다른 함수를 받는 것을 고차함수 <-- 1급함수 : 함수를 변수에 저장가능한 문법
# 콜백함수
# 넘파이, 판다스 API들중(함수, 메서드) 고차함수 형태를 가진것들

num = pd.Series([1, 2, 3, 4])

# def mul(x):
#   if x%2 == 0:
#     return x
#   else:
#     return x*10

# apply() 시리즈 고차함수(메서드)
num.apply(lambda x : x if x%2==0 else x*10)
```

```python
num * 10
```

```python
# 시리즈 개별값에 대해서 맵핑을 메서드
# 맵핑을 할려면 인자값 사전타입

grade = pd.Series(['A', 'B','C', 'D'])
grade_ko = grade.map({'A':'상', 'B':'중', 'C':'하'})
grade_ko
```

```python
import numpy as np
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense
from sklearn.metrics import accuracy_score
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
print(X_train[:2])
print(X_train[-2:])
```

```python
print(len(X_train[0]))
print(len(X_train[1]))
```

```python
for i in range(10):
    print(len(X_train[i]))
```

```python
X_train = pad_sequences(X_train, maxlen=200, padding='pre')
X_test = pad_sequences(X_test, maxlen=200, padding='pre')
```

```python
print(len(X_train[0])) # 1차원 218
print(X_train[1]) # 189
```

```python
print(X_train.shape) # 학습 y_train
print(X_test.shape) # 평기 y_test
```

```python
model = Sequential([
    Input(shape=(200,)),
    Embedding(10000, 32),
    SimpleRNN(10),
    Dense(2, activation='softmax')
])
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(X_train, y_train, batch_size=100, epochs=5, validation_split=0.2)
```

```python
model.evaluate(X_test, y_test)
```

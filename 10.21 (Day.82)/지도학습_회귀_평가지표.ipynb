```python
import math

# 회귀 모델 <- X_test 10개의 입력값
# 예측값 10개
pred = [3, 5, 2, 7, 8, 4, 6, 9, 5, 10]

# 실제 정답 10개
y_test = [2, 5, 3, 8, 7, 4, 5, 10, 6, 9]    # 회귀 : 맞다/틀리다 X -> 각각의 정답-예측값 오차, 정답-평균오차 비교 r2

n = len(y_test)
print(n)
```

```python
# 오차 10개 저장
errors = []

for i in range(n): # n = 10 -> 10번 반복
    d = pred[i] - y_test[i]
    errors.append(d)

print(errors)
```

```python
# MAE : errors에 절댓값을 취한 합을 전체 갯수를 나눈 값

absolute_errors = []

for e in errors:
    absolute_errors.append(abs(e))

print(absolute_errors)
```

```python
MAE = sum(absolute_errors) / n
print('MAE', MAE)
```

```python
# MSE

squared_errors = []

for e in errors:
    squared_errors.append(e**2)

print(squared_errors)
```

```python
MSE = sum(squared_errors) / n
print('MSE : ', MSE)
```

```python
# RMSE : MSE의 루트값
RMSE = math.sqrt(MSE)
print('RMSE :', RMSE)
```

```python
# r2 : 1 - (실제값 - 예측값의 제곱 / 실제값 - 정답 평균값의 제곱) 
# 0 ~ 1 클수록 좋은 모델

# 정답의 평균
y_mean = sum(y_test) / n

t1 = 0
t2 = 2

for i in range(10):
    t1 = t1 + ((y_test[i] - pred[i]) ** 2)
    t2 = t2 + ((y_test[i] - y_mean) ** 2)

r2 = 1 - (t1/t2)

print('r2 :', r2)
```

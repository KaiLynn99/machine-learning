```python
# 1. 데이터 블러오기 및 탐색
# pandas.read_csv() 함수
# DataFrame.Shape, DataFrame.head(), DataFrame.tail, DataFrame.info() 
# DataFrame.describe(), DataFrame.isnull(), DataFrame.isna()

import pandas as pd

train = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_2_train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_2_test.csv')
```

```python
print(train.shape)
print(test.shape)
display(train.head())
display(test.head())
display(train['count'].describe())
```

```python
train.info()
test.info()

print(train.isnull().sum())
```

```python
# 2. X, y 분리
# 인덱싱 DataFrame[], DataFrame.loc[], DataFrame.iloc[]
# DataFrame.drop(axis=)
# 식별자 컬럼(ID) 분리, y(count) 분리 
X = train.drop(columns=['ID', 'count'])
display(X.head(3))

y = train['count']
display(y.head(3))
```

```python
X_test = test.drop(columns=['ID', 'count'])
test_id = test['ID']

display(X_test.head(3))
display(test_id.head(3))
```

```python
# 3. 데이터 분할(train, val, test)
# X, y를 분할
# 학습 : X_train, y_train
# 검증 : X_val, y_val

# 예측 : X_test
# sklearn.model_selection.train_test_split() 함수

from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.3, random_state=1)

print(X_train.shape)
print(X_val.shape)
print(y_train.shape)
print(y_val.shape)
```

```python
# 4. 전처리(비지도 학습)
# 점주형 변수 -> 인코딩
# 연속형 숫자 변수 -> 스케일링(표준화/정규화) -> 모델()KNN, SVM)에 따라 선택
# DataFrame.select_dtypes('object/number).columns -> object/number 컬럼명 리스트
# DataFrame.value_counts()
cat_columns = X_train.select_dtypes('object').columns
num_columns = X_train.select_dtypes('number').columns
print('cat_columns :', cat_columns)
print('num_columns :', num_columns)

# 카테고리(범주형) 컬럼값 -> DataFrame.valu_counts() -> 탐색
# ['holiday', 'workingday', 'weather']
for col in cat_columns:
    print(col)
    display(X_train[col].value_counts())
```

```python
# 전처리 -> 비지도 학습(인코딩, 스케일링, ...) -> 저장
# sklearn.preprocessing.OneHotEncoder, OneHotEncoder.set_output(transfrom='pandas)
# sklearn.preprocessing.StandardScaler, StandardScaler.set_output(transfrom='pandas)

from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler

# sparse_output : 기본값이 numpy배열이 아니다 / handle_unknown : 기본값 error
# encoder, scaler, model 따로 파일 형태로 저장필요
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore').set_output(transform='pandas') 
scaler = StandardScaler().set_output(transform='pandas')
```

```python
# X_train, X_val, X_test의 cat_columns 변수값을 인코딩
# encoder가 필요한 변수 - cat_columns 변수에 저장
X_train_en = encoder.fit_transform(X_train[cat_columns])  # 3개의 컬럼 -> 7개의 컬럼으로 반환
X_val_en = encoder.transform(X_val[cat_columns])
X_test_en = encoder.transform(X_test[cat_columns])

display(X_train_en.head(3))
display(X_val_en.head(3))
display(X_test_en.head(3))

# DataFrame.concat([df1, df2], axis=1) 함수
X_train_pre = pd.concat([X_train_en, X_train[num_columns]], axis=1)
X_val_pre = pd.concat([X_val_en, X_val[num_columns]], axis=1)
X_test_pre = pd.concat([X_test_en, X_test[num_columns]], axis=1)

display(X_train_en.head(3))
display(X_val_en.head(3))
display(X_test_en.head(3))


# SVR모델 사용시 X_train, X_val, X_test의 num_columns 변수값을 인코딩
# scaler가 필요한 변수 - nun_columns 변수에 저장

# 데이터셋을 결합 ex) X_train의 숫자 + X_train_en
```

```python
# 5-1. RamdomForest Model 학습, 검증, 평가
# sklearn.ensemble.RamdomForestRegressor
# sklearn.metrics.mean_absolute_error

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error # 평가지표

model_rf = RandomForestRegressor(random_state=1)
model_rf.fit(X_train_pre, y_train)  
pred_val_rf = model_rf.predict(X_val_pre)   # 평가용 예측치

# 평가용 답안과 비교하여 평가지표(MAE) 구하기
MAE_RF = mean_absolute_error(y_val, pred_val_rf)
print("MAE_RF :", MAE_RF)
```

```python
# 5-2. SVR Model 학습, 검증, 평가

from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler().set_output(transform='pandas')

X_train_sc = scaler.fit_transform(X_train[num_columns])
X_val_sc = scaler.transform(X_val[num_columns])
X_test_sc = scaler.transform(X_test[num_columns])

display(X_train_sc.head(3))
display(X_val_sc.head(3))
display(X_test_sc.head(3))
```

```python
# SVR모델은 데이터셋을 결함
X_train_pre2 = pd.concat([X_train_en, X_train_sc], axis=1)
X_val_pre2 = pd.concat([X_val_en, X_val_sc], axis=1)
X_test_pre2 = pd.concat([X_test_en, X_test_sc], axis=1)

display(X_train_pre2.head(3))
display(X_val_pre2.head(3))
display(X_test_pre2.head(3))

display(X_train_pre2[num_columns].describe())

# sklearn.svm.SVR
model_svr = SVR()
model_svr.fit(X_train_pre2, y_train)
pred_val_svr = model_svr.predict(X_val_pre2)
MAE_SVR = mean_absolute_error(y_val, pred_val_svr)
print("MAE_SVR :", MAE_SVR)
```

```python
# 6. 모델 비교 및 선택
if MAE_RF < MAE_SVR:
    best_model = model_rf
    pred_test = best_model.predict(X_test_pre)
else:
    best_model = model_svr
    pred_test = best_model.predict(X_test_pre2)    
```

```python
# 7. 파이프라인(모델, 인코더, 스케일러) 저장, 스케일러 SVR 모델이 선택되면 스케일러도 저장
# joblib.sump() 함수
import joblib
joblib.dump(encoder, 'my_encoder.joblib')
# joblib.dump(scaler, 'my_scaler.joblib')

joblib.dump(best_model, 'my_model.joblib')
```

```python
# 8. 보고서 -> 결과 파일(csv) 저장
# DataFrame.to_csv() 메서드

result = pd.DataFrame({
    'ID' : test_id,
    '예측치' : pred_test
})

display(result.head())

result.to_csv('result.csv', index=False)
```

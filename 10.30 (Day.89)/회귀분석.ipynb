```python
import pandas as pd
import statsmodels.api as sm

# 학생별 하루 평균 공부·게임·수면 시간과 시험 점수로 만든 샘플 데이터
# 시간 단위는 시간(hour), 시험 점수 단위는 100점 만점
data = pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8, 2, 4, 6, 7],  # 하루 평균 공부시간
    "game_hours":  [5, 4, 4, 3, 3, 2, 2, 1, 5, 4, 3, 1],  # 하루 평균 게임시간
    "sleep_hours": [5, 6, 6, 7, 7, 8, 7, 8, 7, 5, 6, 9],  # 하루 평균 수면시간
    "exam_score":  [42, 50, 55, 61, 68, 75, 79, 88, 48, 58, 71, 84]  # 시험 점수(타깃)
})

# 모델에 넣을 설명변수(X)를 선택
# 공부시간, 게임시간, 수면시간을 사용해 시험 점수를 설명
X = data[["study_hours", "game_hours", "sleep_hours"]]

# 절편(상수항)을 설명변수에 추가
# 절편은 공부·게임·수면 시간이 모두 0일 때의 예측 점수에 해당
X = sm.add_constant(X)
# print('X:', X)
# 예측하려는 값인 타깃 변수(y)를 지정
y = data["exam_score"]

# 최소제곱법(OLS)으로 선형회귀 모델을 적합
# fit()은 주어진 X와 y에 가장 잘 맞는 회귀계수를 계산
model = sm.OLS(y, X).fit()

# 회귀계수, p값, R-squared 등 모델 요약 결과를 출력
print(model.summary())

# coef: 각 시간 변수가 1시간 늘 때 성적이 얼마나 변하는지 나타내는 회귀계수입니다. 다른 시간은 같다고 가정해요.
# P>|t|: 각 변수의 계수가 통계적으로 유의한지 확인합니다. 보통 p값이 0.05 미만이면 유의하다고 해석합니다.
# R-squared: 모델이 시험성적의 차이를 얼마나 설명하는지 나타냅니다. 1에 가까울수록 설명력이 큽니다.
# F-statistic의 Prob (F-statistic): 모델 전체가 통계적으로 유의한지 확인합니다.
```

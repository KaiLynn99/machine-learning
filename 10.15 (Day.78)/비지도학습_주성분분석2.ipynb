```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
# 제공 데이터 NumPy -> Pandas(DF/S)로 변경 -> API 호출
# 제공 데이터 Pandas 데이터셋 -> NumPy로 변경 -> API 호출
# NumPy, Pandas 익숙 -> 배열은 NumPy, DF/S는 Pandas로 호출

from sklearn.datasets import load_breast_cancer # 30개의 변수
from sklearn.preprocessing import StandardScaler # fit -> transform transform
from sklearn.model_selection import train_test_split

from sklearn.decomposition import PCA # 변수 30개 -> PCA 비지도 학습 -> 주성분 변수 2개 -> y 분류
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
```

```python
data = load_breast_cancer()
type(data) # Bunch 사전타입 -> 키/값
```

```python
print(data.feature_names)
print(data.target_names)
```

```python
X = data.data
y = data.target
print(X.shape) # 569행 30변수
print(y.shape) # 569개
```

```python
X[:5]
```

```python
X = pd.DataFrame(X, columns=data.feature_names)
X.head(3)
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1, stratify=y)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model1 = RandomForestClassifier(random_state=1)
model1.fit(X_train, y_train)

pred1 = model1.predict(X_test)
print(pred1) # accuracy_score
print(y_test)
```

```python
print(accuracy_score(y_test, pred1))
print(f1_score(y_test, pred1))
```

```python
X.head(1)
```

```python
scaler = StandardScaler()
```

```python
# 모든 X에서 적용(3개로 분할했다면 X_train, X_val, X_test, ...) fit -> X_train transform - X_val transform - X_test transform

# 이 예제는 2개로 분할
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaler = scaler.transform(X_test)
```

```python
X_train_scaled.round(3)
```

```python
# X의 개수 변경
pca = PCA(n_components=2)

# 변경된 2개 X_train
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaler)
```

```python
print(X_train.shape)
print(X_train_pca.shape)
print(X_test.shape)
print(X_test_pca.shape)
```

```python
model2 = RandomForestClassifier(random_state=1)
model2.fit(X_train_pca, y_train) # 주성분 2개로 변경된 X를 사용

pred2 = model2.predict(X_test_pca) # 주성분 2개로 변경된 X를 사용
print(pred2) # accuracy_score
print(y_test)

print(accuracy_score(y_test, pred2))
print(f1_score(y_test, pred2))
```
